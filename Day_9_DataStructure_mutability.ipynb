## String Immutability
### String can not be modify once we declared

```python
word = "PYTHON"
word[0] = "j"
print (word)
```

```python
word = "PYTHON"
new_word = "J" + word[1:]
print("new:",new_word)
print("old:",word)
```

## Mutable = add, update, remove, insert, extend, sort

```python
#Update
fruits = ["apple", "banana", "orange"]
print(fruits[0])
```

```python
fruits = ["apple", "banana", "orange"]
fruits[0]= "mango"
print(fruits)
```

```python
fruits = ["apple", "banana", "orange"]
fruits.append("mango")    # function methods always use ()
print(fruits)
```

```python
fruits = ["apple", "banana", "orange"]
fruits.append(["mango","cherry"])
print(fruits)
```

```python
fruits = ["apple", "banana", "orange"]
fruits.extend(["mango","cherry"])
print(fruits)
```

```python
# insert
fruits = ["apple", "banana", "orange"]
fruits.insert(1, "mango")
print(fruits)
```

```python
# insert
fruits = ["apple", "banana", "orange"]
fruits.insert(3, "grapes")
print(fruits)
```

```python
# remove
fruits = ["apple", "banana", "orange"]
fruits.remove("banana")
print(fruits)
```

```python
fruits = ["apple", "banana", "orange"]
fruits.remove("orange")
print(fruits)
```

```python
# delete
fruits = ["apple", "banana", "orange"]
del fruits[1]
print(fruits)
```

```python
#pop
fruits = ["apple", "banana", "orange"]
fruits.pop()
print(fruits)
```

```python
fruits = ["apple", "banana", "orange"]
fruits.pop(0)
print(fruits)
```

```python
# sorting
fruits = ["banana", "apple", "orange"]
fruits.sort()
print(fruits)
```

# Dictionary Methods
## dictionary stores the data as key-value pairs

```python
student = {
    "name": "ravi",
    "course":"FDE",
    "age" : 30
}
print(student)
```

```python
student = {
    "name": "ravi",
    "course":"FDE",
    "age" : 30
}
print(student["name"])
```

```python
student = {
    "name": "ravi",
    "course":"FDE",
    "age" : 30,
    "age" : 40
}
print(student)
```

```python
#update
student = {
    "name": "ravi",
    "course":"FDE",
    "age" : 30,
}
student["age"]=40
print(student)
```

```python
student = {
    "name": "ravi",
    "course":"FDE",
    "age" : 30,
}
student["city"]= "Hyderabad"
print(student)
```

```python
student = {
    "name": "ravi",
    "course":"FDE",
    "age" : 30,
}
student["student_name"]= student.pop("name")
print(student)
```
