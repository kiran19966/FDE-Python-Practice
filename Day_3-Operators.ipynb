# Left Shift Right Shift

```python
## Left Shift
x=15
x<<=1
print(x)

#before binary 1111
# after binary 11110
# Left shift (<<): move bits left and add bit position on right 0
```

```python
x=19  #10011
x<<=2 #1001100
print(x)
```

```python
x=21
x<<=3
print(x)
```

```python
## Right Shift
x=15  #1111
x>>=1 #0111
print(x)
```

```python
x=15
x>>=2  #0011
print(x)

#Right Shift = Move bits right and discard the right most bits
```

```python
# Left Shift
15       1111
15 << 1  11110
15 << 2  111100
15 << 3  1111000

# Right Shift
15        1111
15 >> 1   0111
15 >> 2   0011
15 >> 3   0001
```
